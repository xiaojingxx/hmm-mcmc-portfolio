```python
import pandas as pd
from scripts.features.asset_selection import select_assets

df = pd.read_csv("../data/processed/feature_matrix.csv", index_col=0, parse_dates=True)

returns = df[[c for c in df.columns if c.endswith("_ret")]]

final_assets, contrib_scores, dispersion = select_assets(
    returns,
    market_proxy="SPY_ret",
    n_final=15
)

print("Final universe:", final_assets)
```

```python
selected_asset_names = [c.replace("_ret", "") for c in final_assets]
feature_cols = [
    c for c in df.columns
    if any(c == f"{a}_{f}" for a in selected_asset_names for f in ["ret", "vol", "mom"])
]

selected_asset_features_df = df[feature_cols]

selected_asset_features_df.to_csv(f"../data/processed/selected_feature_matrix.csv")
```
