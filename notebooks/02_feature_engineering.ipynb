```python
import pandas as pd

from scripts.features.compute_returns import compute_log_returns
from scripts.features.compute_volatility import rolling_volatility
from scripts.features.technical_indicators import momentum
```

```python
prices = pd.read_csv(
    "../data/raw/prices.csv",
    index_col=0,
    parse_dates=True
)
```

```python
returns = compute_log_returns(prices)
volatility = rolling_volatility(returns)
mom = momentum(prices)
```

```python
mom.tail(30)
```

```python
mom.describe()
```

```python

feature_matrix = pd.concat(
    [
        returns.add_suffix("_ret"),
        volatility.add_suffix("_vol"),
        mom.add_suffix("_mom")
    ],
    axis=1
)
# Keep vol/mom even if NaN, drop only raw returns
ret_cols = [col for col in feature_matrix.columns if col.endswith('_ret')]
feature_matrix = feature_matrix.dropna(subset=ret_cols)  # Only raw returns matter!

print(f"✅ Ret-only dropna: {len(feature_matrix)} rows")
```
