```python
import yfinance as yf
import pandas as pd
```

```python
tickers = ["SPY","QQQ","IWM","DIA","VTI","XLK","XLF","XLE","XLV","XLY","XLP","XLI","XLU","XLRE","XLB","EFA","EEM","EWJ","EWG","EWU","INDA","FXI","BND","IEF","TLT","SHY","LQD","HYG","GLD","SLV","USO","DBC","VNQ","VIXY","TAIL","USMV","MTUM","AAPL","MSFT","NVDA","AMZN","GOOGL","META","TSLA","IBIT","ETHA","COIN","BIL","TIP","SGOV"]
```

```python
prices = yf.download(
    tickers,
    start="2015-01-01",
    end= pd.to_datetime('today').strftime('%Y-%m-%d'),
    auto_adjust=True
)["Close"]

prices.head()
```

```python
prices.to_csv("../data/raw/prices.csv")
```
