```python
import pandas as pd
import numpy as np
from scripts.backtest.backtest_utils import backtest_portfolio

# ---- 1. Align weights and returns on the same dates ----
# weights_df: DataFrame with columns = assets + 'regime', index = dates
# asset_returns_df: DataFrame with same asset columns, daily returns, index = dates

assets = ['AAPL','AMZN','EFA','META','NVDA',
               'SPY','TAIL','TSLA','USMV','VIXY']
ret_cols = {a: f"{a}_ret" for a in assets}

# intersect dates to avoid look-ahead / NA issues
common_idx = weights_rolling.index.intersection(asset_returns.index)

w_t = weights_rolling.loc[common_idx, assets]              # (T, N)
r_t = (
    asset_returns.loc[common_idx, [ret_cols[a] for a in assets]]
    .rename(columns={ret_cols[a]: a for a in assets})
)

# ---- 2. Portfolio returns ----
# dynamic HMM strategy
dyn_ret = (w_t.values * r_t.values).sum(axis=1)            # (T,)

# equal‑weight benchmark (across same assets/dates)
bench_ret = r_t.mean(axis=1).values                        # (T,)

portfolio_dyn = pd.DataFrame(
    {
        'raw_returns': bench_ret,      # equal-weight benchmark
        'scaled_returns': dyn_ret      # regime‑weighted strategy
    },
    index=common_idx
)

raw_ret = portfolio_dyn['raw_returns'].dropna()
scaled_ret = portfolio_dyn['scaled_returns'].dropna()

print("raw len:", len(raw_ret), "scaled len:", len(scaled_ret))

if len(raw_ret) == 0 or len(scaled_ret) == 0:
    raise ValueError("No non‑NaN returns; check date alignment and column mapping.")

raw_perf = backtest_portfolio(raw_ret)
scaled_perf = backtest_portfolio(scaled_ret)


perf_table = pd.DataFrame(
    [raw_perf, scaled_perf],
    index=["Unscaled (Equal‑Weight)", "Regime‑Weighted"]
)

print(perf_table)
```

```python
import numpy as np
import pandas as pd
from scripts.backtest.backtest_utils import backtest_portfolio
from hmmlearn.hmm import GaussianHMM

# --- CONFIG (same as before) ---
window = 272
step   = 21

assets = ['AAPL','AMZN','EFA','META','NVDA','SPY','TAIL','TSLA','USMV','VIXY']
ret_cols = {a: f"{a}_ret" for a in assets}

# Ensure datetime indices
features_df = features_df.copy()
asset_returns = asset_returns.copy()
weights_rolling = weights_rolling.copy()
portfolio_dyn = portfolio_dyn.copy()

for df in [features_df, asset_returns, weights_rolling, portfolio_dyn]:
    df.index = pd.to_datetime(df.index).normalize()

# --- HMM helper with NaN handling ---
def fit_hmm_multistart(X, n_states=2, n_starts=5, n_iter=100, tol=1e-3):
    """
    X: 2D array with no NaNs.
    """
    best_ll, best_model = -np.inf, None
    for _ in range(n_starts):
        m = GaussianHMM(
            n_components=n_states,
            covariance_type="diag",
            n_iter=n_iter,
            tol=tol,
            init_params="stmc"
        )
        m.fit(X)
        ll = m.score(X)
        if ll > best_ll:
            best_ll, best_model = ll, m
    return best_model, best_ll

# -----------------------------
# 1. HMM ROLLING STABILITY (with NaN drop)
# -----------------------------
records = []
feature_dates = features_df.index

for t in range(window, len(features_df), step):
    X_raw = features_df.iloc[t-window:t]
    date_t = feature_dates[t]

    # drop rows with NaNs in this window
    X = X_raw.dropna().values
    if X.shape[0] < 50:  # too few samples? skip this window
        records.append({
            'date': date_t,
            'loglik': np.nan,
            'p11': np.nan,
            'p22': np.nan,
            'mean_state0': np.nan,
            'mean_state1': np.nan,
            'mean_diff': np.nan
        })
        continue

    try:
        model, ll = fit_hmm_multistart(X)
    except ValueError as e:
        # still failed (e.g. numerical issues) – log NaNs and continue
        records.append({
            'date': date_t,
            'loglik': np.nan,
            'p11': np.nan,
            'p22': np.nan,
            'mean_state0': np.nan,
            'mean_state1': np.nan,
            'mean_diff': np.nan
        })
        continue

    trans = model.transmat_
    means = model.means_

    # sort states by mean over features
    sort_idx = np.argsort(means.mean(axis=1))
    means_sorted = means[sort_idx]
    trans_sorted = trans[sort_idx][:, sort_idx]

    records.append({
        'date': date_t,
        'loglik': ll,
        'p11': trans_sorted[0, 0],
        'p22': trans_sorted[1, 1],
        'mean_state0': means_sorted[0].mean(),
        'mean_state1': means_sorted[1].mean(),
        'mean_diff': means_sorted[1].mean() - means_sorted[0].mean()
    })

stability_df = pd.DataFrame(records).set_index('date')

print("\n=== HMM Rolling Stability Summary (NaNs dropped per window) ===")
print(stability_df.describe())

# -----------------------------
# 2. SUBPERIOD PERFORMANCE
# -----------------------------
raw = portfolio_dyn['raw_returns']
scaled = portfolio_dyn['scaled_returns']

periods = [
    ('2016-01-01','2018-12-31'),
    ('2019-01-01','2021-12-31'),
    ('2022-01-01','2025-12-31'),
]

rows = []
for (s,e) in periods:
    r_raw = raw.loc[s:e].dropna()
    r_scaled = scaled.loc[s:e].dropna()
    if len(r_raw) == 0 or len(r_scaled) == 0:
        rows.append({'Period': f'{s}–{e}',
                     'Unscaled_Sharpe': np.nan,
                     'Regime_Sharpe': np.nan})
        continue
    perf_raw = backtest_portfolio(r_raw)
    perf_scaled = backtest_portfolio(r_scaled)
    rows.append({
        'Period': f'{s}–{e}',
        'Unscaled_Sharpe': perf_raw['Sharpe Ratio'],
        'Regime_Sharpe':   perf_scaled['Sharpe Ratio']
    })

stab_perf = pd.DataFrame(rows).set_index('Period')
print("\n=== Subperiod Sharpe Ratios ===")
print(stab_perf)

# -----------------------------
# 3. WEIGHT & REGIME SMOOTHNESS
# -----------------------------
weight_std = weights_rolling[assets].std().sort_values(ascending=False)
turnover = weights_rolling[assets].diff().abs().sum(axis=1)
regime_changes = (weights_rolling['regime'].diff().abs() > 0).sum()

print("\n=== Weight & Regime Smoothness ===")
print("Per-asset weight std (descending):")
print(weight_std)

print("\nTurnover stats (sum |Δw_i| per rebalance):")
print(turnover.describe())

print(f"\nTotal regime switches over sample: {regime_changes}")
```
